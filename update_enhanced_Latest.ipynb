```python

# ==========================
# 1. Distribution Plots (Numeric Features)
# ==========================
numeric_cols = [
    'marks_obtained', 'completion_pct', 'study_hours_per_week',
    'time_spent_hours', 'confidence_score', 'videos_watched',
    'quizzes_attempted', 'peer_interaction_score', 'self_assessment_score'
]

for col in numeric_cols:
    if col in df.columns:
        plt.figure(figsize=(8,5))
        sns.histplot(df[col], bins=30, kde=True, color="skyblue")
        mean_val = df[col].mean()
        median_val = df[col].median()
        
        # Add mean line
        plt.axvline(mean_val, color='red', linestyle='--', label=f"Mean: {mean_val:.1f}")
        plt.title(f"{col.replace('_',' ').title()} Distribution", fontsize=14, fontweight="bold")
        plt.xlabel(col.replace('_',' ').title(), fontsize=12)
        plt.ylabel("Frequency", fontsize=12)
        plt.legend()
        plt.grid(axis='y', linestyle="--", alpha=0.7)
        plt.show()
        
        # Auto-insight
        print(f"📊 Insight: {col.replace('_',' ').title()} ranges from {df[col].min()} to {df[col].max()}, median = {median_val}, mean = {mean_val:.1f}\n")

# ==========================
# 2. Patterns Across Courses, Topics, Learning Style
# ==========================
group_cols = ['course_name', 'topic_name', 'preferred_learning_style']

for col in group_cols:
    if col in df.columns:
        plt.figure(figsize=(10,6))
        sns.violinplot(x=col, y='marks_obtained', data=df, palette="Set2")
        sns.swarmplot(x=col, y='marks_obtained', data=df, color="k", alpha=0.4)
        plt.title(f"Marks Obtained by {col.replace('_',' ').title()}", fontsize=14, fontweight="bold")
        plt.xticks(rotation=45)
        plt.grid(axis='y', linestyle="--", alpha=0.7)
        plt.show()
        
        # Auto-insight
        medians = df.groupby(col)['marks_obtained'].median()
        print(f"📊 Insight: Median marks by {col.replace('_',' ').title()}:\n{medians}\n")

# ==========================
# 3. Segmentation by Demographics
# ==========================
demo_cols = ['age', 'region', 'education_level']

for col in demo_cols:
    if col in df.columns:
        plt.figure(figsize=(10,6))
        sns.boxplot(x=col, y='marks_obtained', data=df, palette="Pastel1")
        sns.stripplot(x=col, y='marks_obtained', data=df, color="k", alpha=0.4)
        plt.title(f"Marks Obtained by {col.replace('_',' ').title()}", fontsize=14, fontweight="bold")
        plt.xticks(rotation=45)
        plt.grid(axis='y', linestyle="--", alpha=0.7)
        plt.show()
        
        # Auto-insight
        medians = df.groupby(col)['marks_obtained'].median()
        print(f"📊 Insight: Median marks by {col.replace('_',' ').title()}:\n{medians}\n")

# ==========================
# 4. Correlation Analysis
# ==========================
plt.figure(figsize=(14,10))
corr = df.corr(numeric_only=True)

# Mask for strong correlation
mask = (corr >= 0.4) | (corr <= -0.4)
sns.heatmap(corr.where(mask), annot=True, cmap="coolwarm", fmt=".2f", cbar=True, linewidths=0.5)
plt.title("Correlation Heatmap of Numeric Features (|corr| >= 0.4)", fontsize=16, fontweight="bold")
plt.show()

# Auto-insight for top correlations
top_corr = corr.unstack().sort_values(ascending=False)
top_corr = top_corr[(top_corr < 1.0) & ((top_corr >= 0.5) | (top_corr <= -0.5))]
print("📊 Strong Correlations (|corr| >= 0.5):")
print(top_corr, "\n")

# ==========================
# 5. Summary Statistics
# ==========================
print("\n🔹 Summary Statistics:")
summary = df.describe(include='all')
print(summary)
```

Student Segmentation / Clustering

This section groups students with similar learning behavior and performance using K-Means clustering.

1. Feature Selection

We selected relevant features for clustering students based on their engagement and performance:

time_spent_hours – Total time spent on the platform

videos_watched – Number of videos watched

quizzes_attempted – Number of quizzes attempted

interactive_exercises_done – Number of exercises completed

forum_posts_count – Participation in forum discussions

marks_obtained – Scores achieved in exams

completion_pct – Course completion percentage

These features are scaled using StandardScaler to normalize the data before clustering.

2. Determine Optimal Number of Clusters (Elbow Method)

The Elbow Method is used to choose the optimal number of clusters 
𝐾
K:

Plot inertia (sum of squared distances) against different K values.

Select the K at which the decrease in inertia slows significantly (the “elbow”).

This ensures clusters are distinct while avoiding overfitting.

3. Apply K-Means Clustering

Chose K = 3 based on the Elbow plot.

K-Means algorithm assigns each student to one of the 3 clusters.

Cluster labels are added to the dataset as df['cluster'].

4. Cluster Summary & Profiling

Number of students per cluster: Gives an overview of cluster size.

Centroids: Show the average values of engagement and performance features for each cluster.

This helps understand the typical behavior of students in each group.

5. Cluster Visualization

A 2D scatter plot visualizes student clusters:

X-axis: time_spent_hours

Y-axis: marks_obtained

Color: Cluster assignment

This plot shows how engagement relates to performance for different student groups.

6. Cluster Insights (Plain English)

For each cluster, we provide:

Number of students

Average marks, study time, and course completion rate

Median age and education level

Most common learning style

Engagement profile description

Interpretation Example:

High average marks + high engagement → Highly engaged & performing well ✅

Low average marks or low engagement → Low engagement or at-risk ⚠️

Deliverables:

Cluster analysis report

Cluster visualizations

Actionable insights for student support and course improvement

```python

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans


# ==========================
# 1. Selecting Features for Clustering
# ==========================
cluster_features = [
    'time_spent_hours', 'videos_watched', 'quizzes_attempted',
    'interactive_exercises_done', 'forum_posts_count', 'marks_obtained',
    'completion_pct'
]

X = df[cluster_features]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ==========================
# 2. Determine Optimal K (Elbow Method)
# ==========================
inertia = []
K_range = range(2, 8)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K_range, inertia, marker='o', color='blue')
plt.title("Elbow Method: Find Optimal Number of Clusters", fontsize=14, fontweight="bold")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()
print("📌 Insight: Choose the K where the elbow occurs (significant drop in inertia).")

# ==========================
# 3. Apply K-Means Clustering
# ==========================
k_opt = 3  # Chosen based on Elbow plot
kmeans = KMeans(n_clusters=k_opt, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)

# ==========================
# 4. Cluster Summary & Profiling
# ==========================
print("📊 Number of students in each cluster:")
print(df['cluster'].value_counts(), "\n")

centroids = pd.DataFrame(scaler.inverse_transform(kmeans.cluster_centers_), columns=cluster_features)
centroids['cluster'] = range(k_opt)
print("📊 Cluster Centroids (average feature values):")
print(centroids, "\n")

# ==========================
# 5. Visualize Clusters (2D)
# ==========================
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='time_spent_hours', y='marks_obtained', 
    hue='cluster', palette='Set1', data=df, s=80
)
plt.title("Student Clusters: Engagement vs Marks", fontsize=14, fontweight="bold")
plt.xlabel("Time Spent Hours")
plt.ylabel("Marks Obtained")
plt.grid(True)
plt.legend(title="Cluster")
plt.show()

# ==========================
# 6. Cluster Insights (Plain English)
# ==========================
for cluster_num in range(k_opt):
    cluster_data = df[df['cluster']==cluster_num]
    avg_marks = cluster_data['marks_obtained'].mean()
    avg_time = cluster_data['time_spent_hours'].mean()
    avg_completion = cluster_data['completion_pct'].mean()
    print(f"📝 Cluster {cluster_num}:")
    print(f" - Number of Students: {len(cluster_data)}")
    print(f" - Average Marks: {avg_marks:.1f}")
    print(f" - Average Study Time (hours): {avg_time:.1f}")
    print(f" - Average Completion Rate: {avg_completion:.2f}")
    
    # Demographic insights
    median_age = cluster_data['age'].median()
    median_edu = cluster_data['education_level'].median()
    common_learning_style = cluster_data['preferred_learning_style'].mode()[0]
    print(f" - Median Age: {median_age}")
    print(f" - Median Education Level: {median_edu}")
    print(f" - Most Common Learning Style: {common_learning_style}")
    
    # Engagement summary
    if avg_marks > df['marks_obtained'].mean():
        engagement_desc = "Highly engaged & performing well ✅"
    else:
        engagement_desc = "Low engagement or at-risk ⚠️"
    print(f" - Engagement Profile: {engagement_desc}\n")
```

Performance & Engagement Analysis

This section helps understand what drives student success or failure on the learning platform.

1. Define Engagement & Performance Metrics

Engagement Metrics:

time_spent_hours – Total time spent on the platform

videos_watched – Number of videos watched

quizzes_attempted – Number of quizzes attempted

interactive_exercises_done – Number of exercises completed

forum_posts_count – Forum participation

study_hours_per_week – Weekly study hours

Performance Metrics:

marks_obtained – Scores achieved in exams

completion_pct – Course completion percentage

These metrics form the basis for correlation analysis and at-risk student identification.

2. Correlation Analysis: Engagement vs Performance

Computes correlation between engagement metrics and performance metrics.

Heatmap visualizes the strength and direction of correlations:

Positive correlation → higher engagement leads to higher performance

Negative correlation → higher engagement leads to lower performance

Key Insights:

Only correlations with |corr| ≥ 0.3 are highlighted for easy understanding.

Example:

Study Hours per Week is positively correlated with Marks Obtained (corr=0.45) → students studying more tend to score higher.

3. Study Time vs Performance

Scatter plot of study_hours_per_week vs marks_obtained.

Regression line (red) shows the trend.

Interpretation:

Students spending more study hours per week generally perform better.

Average study hours and average marks are displayed for reference.

4. Identify At-Risk Students

Thresholds are calculated using median engagement and median marks.

Students below both thresholds are marked as at-risk.

Output:

Number of at-risk students is displayed.

Recommendations include providing extra guidance, tutorials, or personalized follow-ups.

5. Visualize Engagement vs Completion

Scatter plot: quizzes_attempted vs completion_pct.

Color-coded by risk:

Red → At-Risk Students

Green → Safe Students

Interpretation:

Visualizes how engagement (quizzes attempted) impacts completion rates.

Easily identifies low-performing or disengaged students for interventions.

Deliverables:

Correlation heatmaps

Scatter plots with regression trends

List and count of at-risk students

Recommendations for course improvement and student support

```python
# ==========================
# 1. Define Engagement & Performance Metrics
# ==========================
engagement_cols = [
    'time_spent_hours', 'videos_watched', 'quizzes_attempted',
    'interactive_exercises_done', 'forum_posts_count', 'study_hours_per_week'
]

performance_cols = ['marks_obtained', 'completion_pct']

# ==========================
# 2. Correlation Analysis: Engagement vs Performance
# ==========================
plt.figure(figsize=(10,6))
corr_matrix = df[engagement_cols + performance_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation: Engagement Metrics vs Performance", fontsize=14, fontweight="bold")
plt.show()

# Insights for non-technical users
print("📌 Key Insights (Engagement vs Performance):")
for eng in engagement_cols:
    for perf in performance_cols:
        corr_val = corr_matrix.loc[eng, perf]
        if abs(corr_val) >= 0.3:
            direction = "positively" if corr_val > 0 else "negatively"
            print(f" - {eng.replace('_',' ').title()} is {direction} correlated with {perf.replace('_',' ').title()} (corr={corr_val:.2f})")

# ==========================
# 3. Visualize Study Time vs Performance
# ==========================
plt.figure(figsize=(10,6))
sns.scatterplot(x='study_hours_per_week', y='marks_obtained', data=df, alpha=0.7)
sns.regplot(x='study_hours_per_week', y='marks_obtained', data=df, scatter=False, color='red')
plt.title("Study Hours vs Marks Obtained", fontsize=14, fontweight="bold")
plt.xlabel("Study Hours per Week")
plt.ylabel("Marks Obtained")
plt.grid(True)
plt.show()

# Insights
avg_hours = df['study_hours_per_week'].mean()
avg_marks = df['marks_obtained'].mean()
print(f"\n📌 Average study hours per week: {avg_hours:.1f}")
print(f"📌 Average marks obtained: {avg_marks:.1f}")
print("📌 Insight: Students spending more hours tend to score higher (trend shown in red line).")

# ==========================
# 4. Identify At-Risk Students
# ==========================
# Thresholds (can adjust as needed)
low_engagement_threshold = df[engagement_cols].median().mean()  # median engagement value
low_score_threshold = df['marks_obtained'].median()  # median marks

# Students at risk
df['at_risk'] = ((df[engagement_cols].mean(axis=1) < low_engagement_threshold) &
                 (df['marks_obtained'] < low_score_threshold))

at_risk_count = df['at_risk'].sum()
print(f"\n⚠️ Number of at-risk students: {at_risk_count}")
print("⚠️ Recommendation: Provide extra guidance, tutorials, or personalized follow-ups for these students.")

# ==========================
# 5. Visualize Engagement vs Completion
# ==========================
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='quizzes_attempted', y='completion_pct',
    hue='at_risk', palette={True:'red', False:'green'}, data=df, s=80
)
plt.title("Quizzes Attempted vs Completion Rate (Red = At-Risk)", fontsize=14, fontweight="bold")
plt.xlabel("Quizzes Attempted")
plt.ylabel("Completion %")
plt.grid(True)
plt.show()
```

Content & Curriculum Insights

This section helps improve course design by analyzing student performance and learning style effectiveness.

1. Identify Difficult Topics & Courses

Definition of Difficulty:

Topics or courses with low completion rates or low pass rates.

Code Analysis:

topic_summary → Aggregates each topic’s average completion_pct, pass_rate, and marks_obtained.

course_summary → Aggregates each course’s average completion_pct, pass_rate, and marks_obtained.

Output:

Top 5 Difficult Topics → Topics where students struggle the most.

Top 3 Difficult Courses → Courses with lowest overall performance.

📌 These insights help educators focus on topics or courses needing extra attention.

2. Learning Style Effectiveness per Course

Objective: Identify which learning style works best for each course.

Groups data by course_name and preferred_learning_style.

Computes average marks and average completion % for each combination.

Visualizations:

Average Marks by Learning Style per Course

Bar chart comparing student performance across different learning styles (e.g., visual, reading).

Average Completion % by Learning Style per Course

Bar chart showing which learning styles lead to higher course completion.

These plots help instructors adapt teaching methods to maximize student success.

3. Recommend Topics for Extra Support

Criteria for Extra Support:

completion_pct < 60% OR pass_rate < 60%.

```python
# ==========================
# 1. Identify Difficult Topics / Courses
# ==========================
# Difficult = low pass rate or low completion_pct
topic_summary = df.groupby('topic_name').agg({
    'completion_pct':'mean',
    'pass_rate':'mean',
    'marks_obtained':'mean'
}).sort_values(by=['completion_pct', 'pass_rate'])

course_summary = df.groupby('course_name').agg({
    'completion_pct':'mean',
    'pass_rate':'mean',
    'marks_obtained':'mean'
}).sort_values(by=['completion_pct', 'pass_rate'])

# Top 5 difficult topics
difficult_topics = topic_summary.head(5)
print("📌 Top 5 Difficult Topics (low completion & pass rate):")
print(difficult_topics, "\n")

# Top 3 difficult courses
difficult_courses = course_summary.head(3)
print("📌 Top 3 Difficult Courses (low completion & pass rate):")
print(difficult_courses, "\n")

# ==========================
# 2. Learning Style Effectiveness per Course
# ==========================
learning_style_summary = df.groupby(['course_name','preferred_learning_style']).agg({
    'marks_obtained':'mean',
    'completion_pct':'mean'
}).reset_index()

plt.figure(figsize=(12,6))
sns.barplot(x='course_name', y='marks_obtained', hue='preferred_learning_style', data=learning_style_summary)
plt.title("Average Marks by Learning Style per Course", fontsize=14, fontweight="bold")
plt.xlabel("Course Name")
plt.ylabel("Average Marks Obtained")
plt.legend(title="Learning Style")
plt.show()

plt.figure(figsize=(12,6))
sns.barplot(x='course_name', y='completion_pct', hue='preferred_learning_style', data=learning_style_summary)
plt.title("Average Completion % by Learning Style per Course", fontsize=14, fontweight="bold")
plt.xlabel("Course Name")
plt.ylabel("Average Completion %")
plt.legend(title="Learning Style")
plt.show()

# ==========================
# 3. Recommend Topics for Extra Support
# ==========================
# Criteria: completion_pct < 60% or pass_rate < 60%
support_topics = topic_summary[(topic_summary['completion_pct'] < 0.6) | (topic_summary['pass_rate'] < 0.6)]
print("📌 Topics Recommended for Extra Support Modules or Tutorials:")
for idx, row in support_topics.iterrows():
    print(f" - Topic {idx}: Completion {row['completion_pct']:.2f}, Pass Rate {row['pass_rate']:.2f}")
```

Predictive Analytics (Non-ML)

This section uses rule-based statistics to predict trends and identify at-risk students without machine learning models.

1. Define Thresholds for At-Risk Students

Criteria Used:

Low Engagement: Students below median values for at least 3 of these metrics:

time_spent_hours

videos_watched

quizzes_attempted

interactive_exercises_done

forum_posts_count

Low Performance: Students with marks_obtained below the median.

These thresholds are simple, transparent, and easy for non-technical staff to understand.

2. Predict At-Risk Students (Rule-Based)

Implementation:

low_engagement → True if at least 3 engagement metrics are below median

low_performance → True if marks are below median

at_risk → True if both low engagement and low performance

Output:

Total number of at-risk students: ⚠️ Total At-Risk Students: X / Y

Provides a clear indicator of students needing attention.

3. Expected Completion Rates per Course

Aggregates Per Course:

expected_completion_rate → Average completion % of all students in the course

at_risk_percentage → % of students in the course who are at risk

Output Table Example:

Course Name	Expected Completion Rate	At-Risk Percentage
Course A	0.78	15.0%
Course B	0.62	25.0%

This helps instructors anticipate course outcomes and plan interventions.

4. Visualize At-Risk Students

Bar Chart:

Shows number of at-risk vs safe students per course.

Red = At-Risk, Green = Safe

Quick visual for educators to see which courses need urgent attention.

5. Early Warning Alerts

Rule-Based Alerts:

If >20% of students in a course are at risk, the system flags it.

```python

# ==========================
# 1. Define Thresholds for At-Risk Students
# ==========================
# Example rules:
# - Low engagement: below median for time spent, quizzes, videos, exercises
# - Low performance: marks below median
engagement_cols = ['time_spent_hours', 'videos_watched', 'quizzes_attempted', 'interactive_exercises_done', 'forum_posts_count']

# Compute median thresholds
engagement_threshold = df[engagement_cols].median()
marks_threshold = df['marks_obtained'].median()
completion_threshold = df['completion_pct'].median()

# ==========================
# 2. Predict At-Risk Students (Rule-Based)
# ==========================
df['low_engagement'] = df[engagement_cols].lt(engagement_threshold).sum(axis=1) >= 3  # at least 3 metrics below median
df['low_performance'] = df['marks_obtained'] < marks_threshold
df['at_risk'] = df['low_engagement'] & df['low_performance']

at_risk_count = df['at_risk'].sum()
print(f"⚠️ Total At-Risk Students: {at_risk_count} / {len(df)}\n")

# ==========================
# 3. Expected Completion Rates per Course
# ==========================
course_completion = df.groupby('course_name').agg({
    'completion_pct': 'mean',
    'at_risk': 'sum',
    'student_id':'count'
}).reset_index()

course_completion['expected_completion_rate'] = course_completion['completion_pct']
course_completion['at_risk_percentage'] = course_completion['at_risk'] / course_completion['student_id'] * 100

print("📌 Expected Completion Rates & At-Risk Percentage per Course:")
print(course_completion[['course_name','expected_completion_rate','at_risk_percentage']], "\n")

# ==========================
# 4. Visualize At-Risk Students
# ==========================
plt.figure(figsize=(10,6))
sns.countplot(x='course_name', hue='at_risk', data=df, palette={True:'red', False:'green'})
plt.title("At-Risk Students per Course (Red = At-Risk, Green = Safe)", fontsize=14, fontweight="bold")
plt.xlabel("Course Name")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.legend(title="At-Risk Status")
plt.show()

# ==========================
# 5. Early Warning Alerts
# ==========================
print("📌 Early Warning Report:")
for course in course_completion['course_name']:
    course_data = course_completion[course_completion['course_name']==course].iloc[0]
    if course_data['at_risk_percentage'] > 20:  # threshold for alert
        print(f"⚠️ ALERT: {course} has {course_data['at_risk_percentage']:.1f}% at-risk students. Consider intervention.")
```
